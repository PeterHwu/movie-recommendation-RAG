```python
import jieba
from gensim.models import Word2Vec
# Sample Chinese text
text_corpus = [
    "我爱自然语言处理",
    "自然语言处理是一个有趣的领域",
    "机器学习和深度学习是人工智能的两个分支",
    "这段代码演示了如何在Python中使用jieba和Word2Vec"
]
# Step 1: Tokenization using Jieba
tokenized_corpus = [list(jieba.cut(sentence)) for sentence in text_corpus]
# Display Tokenized Sentences
print("Tokenized Corpus:")
for sentence in tokenized_corpus:
    print(sentence)
# Step 2: Train Word2Vec Model
# Define the Word2Vec parameters
model = Word2Vec(sentences=tokenized_corpus, vector_size=100, window=5, min_count=1, workers=4)
# Step 3: Save and Load Model
# Save the model
model.save("word2vec_chinese.model")
# Load the model
# model = Word2Vec.load("word2vec_chinese.model")
# Example: Get vector for a specific word
word = "自然语言"
if word in model.wv:
    word_vector = model.wv[word]
    print(f"Vector for '{word}':")
    print(word_vector)
else:
    print(f"Word '{word}' not found in the vocabulary.")
# Example: Find similar words
similar_words = model.wv.most_similar("自然语言", topn=5)
print("Similar words to '自然语言处理':")
for similar_word, similarity_score in similar_words:
    print(f"{similar_word}: {similarity_score}")
```
